### What is MAE (Mean Absolute Error)?

MAE measures the average magnitude of the errors in a set of predictions, without considering their direction. In simpler terms, it's the average of the absolute differences between predicted and actual values. Every error contributes to MAE in proportion to the absolute value of the error.

**Formula:**

$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$

Where:
- $n$ is the number of data points.
- $y_i$ is the actual value.
- $\hat{y}_i$ is the predicted value.

**Key Characteristics:**
*   **Units:** MAE has the same units as the original data, making it easily interpretable.
*   **Robust to Outliers:** Compared to RMSE, MAE is less sensitive to outliers because it does not square the errors. This means large errors do not disproportionately affect the metric.
*   **Interpretation:** A lower MAE indicates a better model fit. It directly tells you the average absolute difference between your predictions and the true values.

### When to use MAE?
*   When outliers are not considered to be a major problem, and you want a simple, interpretable metric.
*   When the errors are expected to be normally distributed or have a consistent magnitude.

##**MSE(Mean Squared Error)**

image.png

### Advantages of MSE:

*   **Differentiable:** Unlike MAE, MSE is differentiable, which makes it easier to use in optimization algorithms like gradient descent that are commonly used to train machine learning models.
*   **Penalizes Large Errors More:** Because it squares the errors, MSE penalizes larger errors much more significantly than smaller ones. This can be beneficial when large errors are particularly undesirable or costly.
*   **Statistical Properties:** MSE has desirable statistical properties, making it a good choice for many statistical models. It's closely related to variance and is often used in the context of maximum likelihood estimation for normally distributed errors.

### Disadvantages of MSE:

*   **Sensitivity to Outliers:** The squaring of errors means that outliers can have a disproportionately large impact on the MSE value. A few large errors can significantly inflate the MSE, even if the model performs well on most other data points.
*   **Units:** The units of MSE are the square of the original data's units (e.g., if predicting dollars, MSE is in dollars squared). This can make it less intuitive to interpret compared to MAE, which has the same units as the target variable.
*   **Not Always Interpretable:** While mathematically convenient, the squared errors can make it harder to directly understand the typical magnitude of the prediction errors in the context of the original data.

##**RMSE(Root Mean Square Error)**

image.png

### Advantages of RMSE:

*   **Interpretability:** RMSE has the same units as the target variable, making it more interpretable than MSE. It represents the typical magnitude of the prediction errors.
*   **Penalizes Large Errors More:** Similar to MSE, RMSE penalizes larger errors disproportionately due to the squaring operation. This is beneficial when significant errors are particularly undesirable.
*   **Differentiability:** As it is derived from MSE, RMSE is also differentiable, which allows for its use in optimization algorithms like gradient descent during model training.
*   **Commonly Used:** RMSE is a widely used metric in many fields, making it easy to compare model performance across different studies and benchmarks.

### Disadvantages of RMSE:

*   **Sensitivity to Outliers:** Just like MSE, RMSE is highly sensitive to outliers. The squaring of errors means that a few large errors can significantly increase the RMSE value, potentially misrepresenting the overall model performance if those outliers are not representative of the typical data.
*   **Bias Towards Smaller Errors:** Because it penalizes large errors more, a model optimized for RMSE might perform slightly worse on a large number of small errors compared to a model optimized for MAE, which treats all errors linearly.
*   **Not Always Intuitive:** While having the same units as the target variable is an advantage for interpretability, the squaring and square-rooting can sometimes obscure the intuitive understanding of what the 'average error' really means, especially for non-normal error distributions.

### R2 Score (Coefficient of Determination)

The R2 score is a statistical measure that represents the proportion of the variance in the dependent variable that is predictable from the independent variable(s). In simpler terms, it indicates how well the regression model fits the observed data. A higher R2 score indicates a better fit.

**Formula:**

$R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}$

Where:
- $n$ is the number of data points.
- $y_i$ is the actual value.
- $\hat{y}_i$ is the predicted value.
- $\bar{y}$ is the mean of the actual values.

**Key Characteristics:**
*   **Range:** R2 score typically ranges from 0 to 1, but it can be negative if the model performs worse than a simple horizontal line (mean of actual values).
*   **Interpretation:** An R2 of 1 indicates that the regression predictions perfectly fit the data. An R2 of 0 indicates that the model explains none of the variability of the response data around its mean. A negative R2 implies that the model is worse than predicting the mean.
