```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# for categorical Data

## a) Countplot

```python
sns.countplot(df['city'])
```

```python
sns.countplot(df['toss_winner'])
```

## b) pie chart - same as count plot but in the form of percentage

```python
df['neutral_venue'].value_counts().plot(kind='pie', autopct='%0.2f')
```

# for numerical column

##a) Histogram-

```python
sns.histplot(df['result_margin'],bins=10)
```

```python
plt.hist(df['toss_winner'],bins=50)
```

##b) Distplot- used to show distribution of data same as hist plot but in the form of probability in y axis using pdf(probability distribution function)

```python
df['result_margin'].skew()
# +ve value means righ skewness
# -ve means left skewed
# 0 means symmetry
```

```python
sns.distplot(df['result_margin'],kde=True,hist=True,rug=False)
```

## c) Box Plot- used to detect the potential outliers in the graph

```python
sns.boxplot(df['toss_winner'])
```
