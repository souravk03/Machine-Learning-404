## One hot encoding

### DUmmy variable are created after One hot encoding and there is some collinerity between the new dummy variable created and to avoid this **Dummy variable trap or Multicollinearity**, we have to drop the first dummy column after One hot encoding

```python
df['owner'].value_counts()
```

```python
df['fuel'].value_counts()
```

# One Hot Encoding using pandas

```python
pd.get_dummies(df,columns=['fuel','owner'])
```

#### as you can see from the above dataframe there are 12 columns are created where since there are 4 categories in the fuel 4 new dummy vehicle are created for fuel and other 4 are from owners. now there is problem of **Multicollinerity arises as there is some mathematical relation between the input data to remove these problem we have to delete the first column of dummy variable**

# K-1 Encoding where K is categories

```python
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True,dtype='int32') #default dummy variables are bool
```
