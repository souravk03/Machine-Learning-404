```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Datasets/train.csv')
```

```python
df
```

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
df.head()
```

```python
# Step 1 -> train/test/split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Survived']),
                                                 df['Survived'],
                                                 test_size=0.2,
                                                random_state=42)
```

```python
X_train.head(2)
```

```python
y_train.head()
```

```python
df.isnull().sum()
```

```python
# Applying imputation

si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent')

X_train_age_imputed = si_age.fit_transform(X_train[['Age']])
X_train_embarked_imputed = si_embarked.fit_transform(X_train[['Embarked']])

X_test_age_imputed = si_age.transform(X_test[['Age']])
X_test_embarked_imputed = si_embarked.transform(X_test[['Embarked']])
```

```python
X_train_embarked
```

```python
# one hot encoding Sex and Embarked

ohe_sex = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
ohe_embarked = OneHotEncoder(sparse_output=False,handle_unknown='ignore')

# Fit and transform Sex and Embarked using the imputed data
X_train_sex_encoded = ohe_sex.fit_transform(X_train[['Sex']])
X_train_embarked_encoded = ohe_embarked.fit_transform(X_train_embarked_imputed)

X_test_sex_encoded = ohe_sex.transform(X_test[['Sex']])
X_test_embarked_encoded = ohe_embarked.transform(X_test_embarked_imputed)
```

```python
X_train_embarked
```

```python
X_train.head(2)
```

```python
X_train_rem = X_train.drop(columns=['Sex','Age','Embarked'])
X_test_rem = X_test.drop(columns=['Sex','Age','Embarked'])
```

```python
X_train_transformed = np.concatenate((X_train_rem,X_train_age_imputed,X_train_sex_encoded,X_train_embarked_encoded),axis=1)
X_test_transformed = np.concatenate((X_test_rem,X_test_age_imputed,X_test_sex_encoded,X_test_embarked_encoded),axis=1)
```

```python
X_test_transformed.shape
```

```python
clf = DecisionTreeClassifier()
clf.fit(X_train_transformed,y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
ohe_sex = pickle.load(open('models/ohe_sex.pkl','rb'))
ohe_embarked = pickle.load(open('models/ohe_embarked.pkl','rb'))
clf = pickle.load(open('models/clf.pkl','rb'))
```

```python
# Assume user input
# Pclass/gender/age/SibSp/Parch/Fare/Embarked
test_input = np.array([2, 'male', 31.0, 0, 0, 10.5, 'S'],dtype=object).reshape(1,7)
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
test_input_sex
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:,-1].reshape(1,1))
test_input_embarked
```

```python
test_input_age_imputed = si_age.transform(test_input[:,2].reshape(1,1))
test_input_transformed = np.concatenate((test_input[:,[0,3,4,5]],test_input_age_imputed,test_input_sex,test_input_embarked),axis=1)
test_input_transformed.shape
```

```python
# To match the 13 features expected by the model, we infer the following structure:
# 1. Pclass one-hot encoded (3 features)
# 2. SibSp, Parch, Fare (3 features)
# 3. Age (imputed, 1 feature)
# 4. Sex one-hot encoded (2 features)
# 5. Embarked one-hot encoded (3 features)
# 6. One additional feature (placeholder, 1 feature)

# 1. Properly initialize and fit ohe_pclass based on training data's Pclass values
# Assuming Pclass categories are 1, 2, 3 from the original dataset
ohe_pclass = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
ohe_pclass.fit(np.array([1, 2, 3]).reshape(-1, 1)) # Fit on all possible Pclass values

# Transform test_input Pclass using the fitted encoder
test_input_pclass_encoded = ohe_pclass.transform(test_input[:,0].reshape(1,1))

# 2. The original numerical features (SibSp, Parch, Fare) without Pclass
test_input_numerical_rem = test_input[:,[3,4,5]]

# Recalculate test_input_transformed with the correct feature structure
test_input_transformed = np.concatenate(
    (test_input_numerical_rem,       # 3 features (SibSp, Parch, Fare)
     test_input_age_imputed,         # 1 feature (Age)
     test_input_pclass_encoded,      # 3 features (Pclass one-hot encoded)
     test_input_sex,                 # 2 features (Sex one-hot encoded)
     test_input_embarked,            # 3 features (Embarked one-hot encoded)
     np.zeros((test_input.shape[0], 1)) # 1 feature (Placeholder for the inferred 13th feature)
    ), axis=1)

print(f"Shape of test_input_transformed: {test_input_transformed.shape}")
clf.predict(test_input_transformed)
```
