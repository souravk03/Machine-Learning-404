#Working with Dates

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

```python
#1. Extract year
date['date_year'] = date['date'].dt.year

date.sample(5)
```

```python
#2. Extract Month
date['date_month_no'] = date['date'].dt.month

date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

```python
#Extract Days
date['date_day'] = date['date'].dt.day

date.head()
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek

date.head()
```

```python
# day of week - name

date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
#Extract week of the year
date['date_week'] = date['date'].dt.isocalendar().week

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
##Extract Quarter
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
#Extract Semester
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
#Extract Time elapsed between dates
import datetime

today = datetime.datetime.today()

today#
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed

# Calculate total months from a reference point for both today and each date in the 'date' column
today_months = today.year * 12 + today.month
date_months = date['date'].dt.year * 12 + date['date'].dt.month

# Calculate the difference in months
months_passed = today_months - date_months

# Round the result
np.round(months_passed, 0)
```
