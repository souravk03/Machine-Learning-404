#Multivariate imputer using KNN imputer.

image.png

The K-Nearest Neighbors (KNN) Imputer is a method used to fill in missing values (impute) in a dataset. Instead of just replacing missing values with a mean or median, it uses the values from 'similar' data points to make a more intelligent guess.

Here's how it generally works:

Find Neighbors: For a data point with a missing value, KNN Imputer looks for 'k' other data points that are most similar to it based on the features that do exist.
Impute: Once these 'k' nearest neighbors are found, their values for the feature with the missing data are used to estimate the missing value. This could be by taking the mean, median, or a weighted average of those neighbors' values.
In short, KNN Imputer fills missing data by finding the 'k' most similar existing data points and using their values to estimate the missing one.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer,KNNImputer
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Datasets/train.csv')[['Age','Pclass','Fare','Survived']]
```

```python
df.head()
```

```python
df.isnull().mean() * 100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
X_train.head()
```

```python
knn = KNNImputer(n_neighbors=3,weights='distance')

X_train_trf = knn.fit_transform(X_train)
X_test_trf = knn.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(X_train_trf,y_train)

y_pred = lr.predict(X_test_trf)

accuracy_score(y_test,y_pred)
```

```python
# Comparision with Simple Imputer --> mean

si = SimpleImputer()

X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(X_train_trf2,y_train)

y_pred2 = lr.predict(X_test_trf2)

accuracy_score(y_test,y_pred2)
```
