```python
train=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/train.csv')
train.head()
```

```python
sns.barplot(data=train,x='Sex',y='Survived',hue='Pclass')
```

#3. Boxplot (Numerical- categorical)
x axis - categorical , y- numerical

```python
sns.boxplot(data=train,x='Survived',y='Age', hue='Sex')
```

#4. Distplot (Numerical- categorical)
x axis - categorical , y- numerical

```python
sns.distplot(train[train['Survived']==0]['Age'],hist=False)
sns.distplot(train[train['Survived']==1]['Age'],hist=False)
```

#5. HeatMap (Categorical - Categorical)

```python
train.head()
```

```python
sns.heatmap(pd.crosstab(train['Pclass'],train['Survived']))
```

```python
train.groupby('Embarked')['Survived'].mean()*100
```
