#Feature Transformer

Encoding Categorical Values

For Ordinal categorical input data we use ordinal encoder

For Output Ordinal data we use Label Encoder

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/customer.csv')
```

```python
df=df.iloc[:,2:]
df
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
```

```python
Oe=OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
xtrain,xtest,ytrain,ytest=train_test_split(df.iloc[:,0:2],df.iloc[:,-1], test_size=0.2)
```

```python
Oe.fit(xtrain)
```

```python
xtrain=Oe.transform(xtrain)
xtest=Oe.transform(xtest)
```

```python
xtrain
```

# Label Encoder is used when there is already Ordinal encoder because label encoder is used only for target variable is it is ordinal.

```python
from sklearn.preprocessing import LabelEncoder
```

```python
Le=LabelEncoder()
```

```python
Le.fit(ytrain)
```

```python
ytrain=Le.transform(ytrain)
ytest=Le.transform(ytest)
```
